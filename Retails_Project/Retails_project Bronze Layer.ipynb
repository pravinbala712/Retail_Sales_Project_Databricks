# **Bronze Layer**

```python
customer_df=spark.read.csv("/Volumes/retails_project/raw_data/raw_file/retail_customers_messy.csv", header=True, inferSchema=True)

customer_df.display()
```

```python
orders_df=spark.read.csv("/Volumes/retails_project/raw_data/raw_file/retail_orders_messy.csv", header=True, inferSchema=True)

orders_df.display()
```

```python
from pyspark.sql.functions import current_timestamp
customer_bronze_df=customer_df.withColumn("ingest_time", current_timestamp())

customer_bronze_df.write.format('delta').mode('overwrite').saveAsTable("retails_project.bronze.customers_bronze")
display(customer_bronze_df)
```

```python
orders_bronze_df=orders_df.withColumn("ingest_time", current_timestamp())
orders_bronze_df.write.format('delta').mode('overwrite').saveAsTable("retails_project.bronze.orders_bronze")
display(orders_bronze_df)
```
